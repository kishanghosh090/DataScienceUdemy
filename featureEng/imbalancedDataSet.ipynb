```python
import numpy as np
import pandas as pd

np.random.seed(123)

n_samples = 100
class_0_ratio = 0.9

n_class_0 = int(n_samples * class_0_ratio)
n_class_1 = n_samples - n_class_0
```

```python
n_class_0,n_class_1
```

```python
#@ Create nt dataframe with imbalanced dataset

class_0 = pd.DataFrame({
    'feature_1': np.random.normal(loc=0, scale=1, size=n_class_0),
    'feature_2': np.random.normal(loc=0, scale=1, size=n_class_0),
    'target': [0] * n_class_0
})

class_1 = pd.DataFrame({
    'feature_1': np.random.normal(loc=3, scale=1, size=n_class_1),
    'feature_2': np.random.normal(loc=3, scale=1, size=n_class_1),
    'target': [1] * n_class_1
})
```
