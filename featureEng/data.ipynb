## inpution missing values
1- mean value impution

```python
sns.displot(df['age'])
```

```python
df['age_mean'] = df['age'].fillna(df['age'].mean())
```

```python
df
```

### mean imputation works well when we have normally distrubution data

```python
## Median value impution if we have outliers in the dataset


df['age_median'] = df['age'].fillna(df['age'].median())
```

```python
df
```

```python
### Mode imputaion technique categorical

df['embarked'].unique()
```

```python
df['embarked_mode'] = df.fillna()
```

```python
df['embarked_mode'] = df['embarked'].fillna(df[df['embarked'].notna()]['embarked'].mode()[0])
```
