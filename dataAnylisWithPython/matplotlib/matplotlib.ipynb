```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

plt.plot(x, y)
plt.show()
```

```python
plt.plot(x, y, 'ro')
plt.xlabel('X label')
plt.ylabel('Y label')
plt.title('Title')
plt.legend()
plt.show()
```

```python
x=[x for x in range(10)]
y=[x for x in range(10,0,-1)]


# create line plot

plt.plot(x,y)
plt.xlabel('X label')
plt.ylabel('Y label')
plt.title('basic line plot')
# show plot
plt.show()
```

```python
##  create custom plot

plt.plot(
    x,
    y,
    color='red',
    # linestyle='dashed',
    linewidth=3,
    marker='o',
    markerfacecolor='blue',
    markersize=12,
)
plt.xlabel('X label')
plt.ylabel('Y label')
plt.title('custom line plot')

# show plot
plt.show()
```
